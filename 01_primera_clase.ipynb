```python
import numpy as np
import pandas as pd

np.random.seed(32)
```

```python
tamaños = [10, 100, 1000, 10000, 100000]
E_teorica = 3.5

n_sims = []
esperanzas = []
errores = []

for i in tamaños:
    tiradas = np.random.randint(1, 7, size=i)
    media = tiradas.mean()

    n_sims.append(i)
    esperanzas.append(media)
    errores.append(abs(media - E_teorica))
```

```python
tabla = pd.DataFrame({
    'N': n_sims,
    'Esperanza simulada': esperanzas,
    'Esperanza teórica': E_teorica,
    'Error absoluto': errores
})

tabla
```

### Caso Hipotetico
Tirar 3 dados consecutivamente, y quiero apostar 10,000 pesos a que cada tiro sale un numero mayor que el anterior, cual es la probabilidad de que gane la apuesta? usando Montecarlo


```python
np.random.seed(32)

N = 100000
victorias = 0

for i in range(N):
    d1 = np.random.randint(1, 7)
    d2 = np.random.randint(1, 7)
    d3 = np.random.randint(1, 7)
    if d1 < d2 < d3:
        victorias += 1

prob = victorias / N
print(f"P(ganar) = {prob}")
```
